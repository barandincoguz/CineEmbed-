# CineEmbed — 00 Drive Setup (one-time)

Run this notebook ONCE before running any training notebook on Colab.
It mounts Google Drive and verifies the expected folder structure.

## Required Drive structure (set up manually before running this notebook)

```
MyDrive/cineembed/
├── repo/
│   ├── pyproject.toml
│   ├── src/cineembed/
│   └── notebooks/
└── artifacts/
    ├── feature_matrix.npz
    ├── movies_eda_final.csv
    └── feature_metadata.json (and other EDA outputs)
```

### How to upload your project to Drive (one-time, ~5 min)

**Option A — Google Drive desktop app (recommended)**:
1. Install Google Drive for desktop on your Mac.
2. In Finder, open the synced Drive folder.
3. Create folder `cineembed/repo` and copy these from your project:
   - `pyproject.toml`
   - `pyrightconfig.json`
   - `src/`
   - `notebooks/`
   - `tests/` (optional, helpful if you want to run pytest in Colab)
4. Create folder `cineembed/artifacts` and copy these from your local `artifacts/`:
   - `feature_matrix.npz`
   - `movies_eda_final.csv`
   - `feature_metadata.json`
   - `pipeline_version.json`
   - `director_profile_metadata.json`
5. Wait for Drive to sync (small status icon turns green).

**Option B — drive.google.com web UI**:
1. Go to https://drive.google.com
2. Create the folder structure manually.
3. Drag-drop the listed files/folders.

````python
import os, sys
from pathlib import Path

IN_COLAB = 'google.colab' in sys.modules
assert IN_COLAB, "This notebook is for Colab only — run on https://colab.research.google.com"

from google.colab import drive
drive.mount('/content/drive')

DRIVE_BASE = Path('/content/drive/MyDrive/cineembed')
REPO_ROOT  = DRIVE_BASE / 'repo'
ARTIFACTS  = DRIVE_BASE / 'artifacts'

print(f"Drive base: {DRIVE_BASE}")
print(f"Drive base exists: {DRIVE_BASE.exists()}")
print(f"Repo on Drive:     {REPO_ROOT.exists()}")
print(f"Artifacts on Drive: {ARTIFACTS.exists()}")
````

````python
required = {
    'repo files': [
        REPO_ROOT / 'pyproject.toml',
        REPO_ROOT / 'src' / 'cineembed' / '__init__.py',
        REPO_ROOT / 'src' / 'cineembed' / 'data.py',
        REPO_ROOT / 'src' / 'cineembed' / 'losses.py',
        REPO_ROOT / 'src' / 'cineembed' / 'backbone.py',
        REPO_ROOT / 'src' / 'cineembed' / 'heads.py',
        REPO_ROOT / 'src' / 'cineembed' / 'train.py',
        REPO_ROOT / 'src' / 'cineembed' / 'eval.py',
    ],
    'artifacts': [
        ARTIFACTS / 'feature_matrix.npz',
        ARTIFACTS / 'movies_eda_final.csv',
        ARTIFACTS / 'feature_metadata.json',
        ARTIFACTS / 'pipeline_version.json',
    ],
}

all_ok = True
for category, paths in required.items():
    print(f"\n[{category}]")
    for p in paths:
        ok = p.exists()
        all_ok &= ok
        marker = '\u2713' if ok else '\u2717'
        print(f"  {marker} {p}")

# Create models/ and eval/ dirs in artifacts
(ARTIFACTS / 'models').mkdir(exist_ok=True)
(ARTIFACTS / 'eval').mkdir(exist_ok=True)
print(f"\nCreated/verified {ARTIFACTS / 'models'} and {ARTIFACTS / 'eval'}")

if all_ok:
    print(f"\n\u2705 Drive setup OK. Open 02_train_ae.ipynb next.")
else:
    print(f"\n\u274c Missing files — see above. Re-check the upload.")
````
